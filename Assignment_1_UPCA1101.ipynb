```python
#Skipping specific numbers using pass
for i in range(1,20):
  if i !=5:
    pass
    print(i)
```

```python
#Break #stop if conditions satisfies and print next numbers
for num in range(1,20):
  if num == 11:
    break
    print(num)
  else:
    print(num)
```

```python
#Continue -- Check the conditions and continue to print
for a in range(1,70):
  if a==50:
    continue
    print(a)
  else:
    print(a)
```

# Write a Python program to find the largest number in a list.

```python
#method 1 : Using for loop with two variables:  1 hold the max of range (1,21) = 20 and another hold the list of ranges (1,2,3... 20)
for i in range(1,21):
  largest = max(range(1,21))   #Assigning max(range) to largest variable (20)
  if i >= largest:             # i has list of values from 1 to 20 and it compare with largest (20)
    print(i)                   # if i > = largest(20) ==> Print i value
```

```python
#method 2 : Single line command using max() aggregate functions
large_num = max(range(1,30))
print("Largest number in this range :",large_num)
```

# **Write a Python program to count the frequency of elements in a list.**

```python
#using list -- But it would print the duplicate on each iterations.
Lst_1 = [1,2,3,100,55,90,1,0,2,65,89,45,55,60,43,23,11,100,9,8,3]
for i in Lst_1:
      print(i, "appears", Lst_1.count(i)," times")
```

```python
#using list with counter list. It holds the value on each iteration

Lst_1 = [1,2,3,100,55,90,1,0,2,65,89,45,55,60,43,23,11,100,9,8,3]
counter =[]
for i in Lst_1:
  if i not in counter:
      print(i, "appears", Lst_1.count(i)," times")
      counter.append(i)
```

# Create a dictionary that stores student names as keys and marks as values, then find the student with the highest marks.

```python
students = {
    "Rathish": 75,
    "Vasanth": 92,
    "Arjun": 78,
    "John": 95
}

max_mark = max(students.values()) # Here i will take maximum mark as value (95) and store it in max_mark variable.

for name, mark in students.items():   # using for loop i have used items() to store name and mark in for loop
    if mark == max_mark:              # i am comparing 1st value with 95 and so on . Till conditions satisfies.
                                      # Print name and mark since mark == max_mark
        print("Student Name =", name)
        print("Maximum Mark =", mark)
```

# **Implement a while loop to print the Fibonacci series up to a given number.**

```python
n = int(input("Enter number :")) #Get max range . example: 10
#default value assignment
a = 0
b = 1
cnt =0


while cnt < n:  #if 0 < 10
    print(a)         # 0,1,....
    c = a + b        #c = 0 + 1 ==> 1
    a = b            # a=b a==> 1
    b = c            # b=c b==> 1
    cnt+=1           #increment by 1 ==> cnt = 1
```

```python
#type 2 Using for loop
n = int(input("Enter number :")) #Get max range . example: 4
#default value assignment
a = 0
b = 1

for i in range(n):   #range from 0 to 4
    print(a)         # 0
    c = a + b        #c = 0 +1 ==> 1
    a = b            # a=b a==> 1
    b = c            # b=c b==> 1
```
