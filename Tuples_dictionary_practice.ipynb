# ***Dictionaries***

```python
#Create dictionary
dict1 ={'Name' : 'Raj',
      'Age':  '30',
      'Job': 'Merchent'}
```

```python
dict1
```

```python
dict1['Name']
```

```python
#updating values in dict1
dict1['Age'] = 32
```

```python
dict1
```

```python
#Adding key value pair
dict1['Country']='London'
dict1['Pincode']='10001'
```

```python
dict1
```

```python
#Removing key value pair using "del"

del dict1['Pincode']
```

```python
dict1
```

# **Functions in Dictionary**

```python
print(dict1.keys())
print(dict1.values())
print(dict1.items())
```

# **Loops**

```python
dict1
```

```python
#print key and values type 1
for x in dict1:
  print(x,":", dict1[x])
```

```python
# using items type 2
for keys, value  in dict1.items():
  print (keys,":", value)
```
