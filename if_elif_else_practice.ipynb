***If else - Comparing 2 numbers***

```python
a=10
b=10
if a > b:
  print('A is bigger than B')
elif a < b:
  print('B is bigger than A')
else:
  print ('A is equal to B')
```

***Nested if - Age- License***

```python
age=18
license = False
if age >=18:
  print('You are eligible to drive')
  if license:
    print ('You can drive Car / 2 wheeler')
  else:
    print('You should get license')
else:
  print('You are not eligible')
```

***Nested if - Bills***

```python
bill=1900
silver_member=False
gold_member=True
platinum_member=False
if bill >=1500:
  if silver_member:
    print('10% Discount')
  elif gold_member:
      print('40% Discount')
  elif platinum_member:
      print('50% Discount')
  else:
    print('No discount - Kindly apply for membership')
else:
  print('Not eligible for discount')
```
