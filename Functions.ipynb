***Playing with parameters***

```python
#pass 2 numbers , addition of 2 numbers
def add_two_numbers(val1,val2):
  print("User given number value1",val1)
  print("User given number value2",val2)
  c=val1+val2
  print("Sum of two numbers",c)

```

```python
add_two_numbers(10,20) #Positional arguments
```

```python
add_two_numbers(20,10)
```

```python
#Keyword argument
add_two_numbers(val2=20,val1=10)
```

```python
#Default parameter
def add_two_numbers(val1,val2=10):
  print("User given number value1",val1)
  print("User given number value2",val2)
  c=val1+val2
  print("Sum of two numbers",c)
```

```python
add_two_numbers(20)
```

```python
#40 override the default param value
add_two_numbers(20,40)
```

**return keyword**

```python
def mul_of_two_numbers(num1,num2):
  print("User given number value1",num1)
  print("User given number value2",num2)
  c=num1*num2
  print("Sum of two numbers",c)
  return c
```

```python
result = mul_of_two_numbers(10,20)
print(result)
```

Assessment
Write a function that returns the total cost of purchase.
Parameters require are price and quantity

```python
def purchase(p1,q1):
  print("Unit price of product :",p1)
  print("Quantities required :",q1)
  cost=p1*q1
  print("Total cost of purchase :",cost)
  return cost
```

```python
cost = purchase(200,5)
print(cost)
```
