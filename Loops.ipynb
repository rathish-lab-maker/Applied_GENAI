While Loop

```python
#print values from 1 to 20
# initializtion, Condition , increment/decrement
c=1
while c<=20:
  print(c)
  c=c+1
```

```python
#print value from 20 to 1
a=20
while a>=1: # 20 >=1
  print(a)
  a-=1
```

For loop with range

```python
#print values from 1 to 10
for i in range(1,11,1):
  print(i)
```

```python
#print all the even numbers between 1 to 20 (range)
for a in range(2,21,2):
  print(a)
```

```python
#print 0 to 5 with stop value
for i in range(6):
  print(i)
```

```python
#print numbers from 22 to 11
for i in range(22,10,-1):
  print(i)
```

For Loop with conditional statments

```python
for a in range(1,11):
  print(a)
```

```python
for a in range(1,11):
  print(a)
```

```python
if (10 % 2 == 0):
  print("even")
else:
  print("odd")
```

```python
for a in range(1,11):
  #print(a)
  if (a % 2 == 0):
    print("even")
  else:
    print("odd")
```

```python
#Print from 1 to 10 but i dont need 5
for i in range(1,11):
  if i !=5:
    print(i)
```

```python
for i in range(1,11):
  if i == 5:
    pass #nothing will print
  else:
    print(i)
```

```python
for i in range(1,11):
  if i == 5:
    continue #Skip 5
  else:
    print(i)
```

```python
#print 5,15,20,25,30
for i in range(5,31,5):
  if i==10:
    continue #Skip
  print(i)
```

```python
for i in range(1,11):
  if i == 5:
    break #Stop
  else:
    print(i)
```

```python
#Run loop from 1 to 100 but it need to stop at 20 Type 1 using condition ==21
for a in range(1,101):
  if a ==21: #check the value and print / break
    break
  print(a)
```

```python
#Run loop from 1 to 100 but it need to stop at 20 Type 2 by using condition ==20
for a in range(1,101):
  print(a) # Print value and check the conditions below
  if a ==20:
    break
```
